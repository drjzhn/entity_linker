```python
from base.matcher import build_string_matcher, find_matches
import json
import os
import glob
```

```python
concepts_path = "concepts/*.json"
concepts = []

for json_file in glob.glob(concepts_path):
    try:
        with open(json_file, 'r') as f:
            concept = json.load(f)
            concepts.append(concept)
    except Exception as e:
        print(f"Error loading {json_file}: {e}")

matcher = build_string_matcher(concepts)
print(f"Loaded {len(matcher)} terms from {len(concepts)} concepts")
```

```python
test_text = """
The patient has a history of Alzheimer's Disease and recently showed symptoms of Parkinson's Disease.
"""

matches = find_matches(test_text, matcher)
matches
```
